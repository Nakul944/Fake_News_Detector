# punctuations, number etc in statements

```python
train["has_number"] = train["statement"].str.contains(r"\d", regex=True)

train["has_number"].value_counts()
```

```python
train.loc[
    train["has_number"],
    ["statement", "label"]
].head(10)
```

```python
train["has_url"] = train["statement"].str.contains(
    r"http|www\.",
    case=False,
    regex=True
)

train["has_url"].value_counts()
```

```python
train["uppercase_count"] = train["statement"].apply(
    lambda x: sum(1 for c in x if c.isupper())
)

train["uppercase_count"].describe()
```

```python
import string

train["punctuation_count"] = train["statement"].apply(
    lambda x: sum(1 for c in x if c in string.punctuation)
)

train["punctuation_count"].describe()
```

# Cross-Split Overlap

```python
train_speaker_list = train["speaker"].unique()     
type(train_speaker_list)
len(train_speaker_list)
```

```python
common_train_test = []
common_train_valid = []
for speaker in train_speaker_list:
    if speaker in test["speaker"].values : 
        common_train_test.append(speaker)

    if speaker in valid["speaker"].values : 
        common_train_valid.append(speaker)
```

```python
print(len(common_train_test))
print(len(common_train_valid))
```

```python
for i in common_train_test:
    print(i)
```

```python
print("Unique train speakers:", len(train_speaker_list))
print("Unique test speakers:", len(test["speaker"].unique()))
print("Unique valid speakers:", len(valid["speaker"].unique()))

print(
    "Test speakers seen in train:",
    len(common_train_test) / len(test["speaker"].unique()) * 100
)

print(
    "Validation speakers seen in train:",
    len(common_train_valid) / len(valid["speaker"].unique()) * 100
)
```

```python
train_statements = set(train["statement"])
test_statements = set(test["statement"])
valid_statements = set(valid["statement"])

common_train_test_statements = train_statements & test_statements
common_train_valid_statements = train_statements & valid_statements

print(
    "Statements shared with test:",
    len(common_train_test_statements)
)

print(
    "Statements shared with validation:",
    len(common_train_valid_statements)
)
```

```python
for stmt in common_train_test_statements:
    print("\nStatement:", stmt)
    print("Train labels:", train.loc[train["statement"] == stmt, "label"].unique())
    print("Test labels :", test.loc[test["statement"] == stmt, "label"].unique())
```

```python
for stmt in common_train_valid_statements:
    print("\nStatement:", stmt)
    print("Train labels:", train.loc[train["statement"] == stmt, "label"].unique())
    print("Test labels :", valid.loc[valid["statement"] == stmt, "label"].unique())
```

# Truth-History analysis

```python
history_cols = [
    "barely_true_counts",
    "false_counts",
    "half_true_counts",
    "mostly_true_counts",
    "pants_on_fire_counts"
]
```

```python
train[history_cols].isnull().sum()
```

```python
train[history_cols].describe()
```

```python
for col in history_cols:
    print(col, (train[col] == 0).sum())
```

```python
(train[history_cols] < 0).sum()
```

```python
train["total_history"] = train[history_cols].sum(axis=1)

train["total_history"].describe()
```

```python
train.nlargest(493, "total_history")[
    ["speaker"] + history_cols + ["total_history"]
]
```

```python
(train["total_history"] == 473).value_counts()
```

```python
train.loc[
    train["total_history"] == train["total_history"].max(),
    "speaker"
].value_counts()
```

```python
speaker_history_variation = (
    train.groupby("speaker")[history_cols]
    .nunique()
)

speaker_history_variation.max()
```

```python
varying_speakers = speaker_history_variation[
    (speaker_history_variation > 1).any(axis=1)
]

varying_speakers
```

```python
speaker = varying_speakers.index[0]

train.loc[
    train["speaker"] == speaker,
    ["speaker"] + history_cols
].drop_duplicates()
```
